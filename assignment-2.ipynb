```python
#Running Sum of 1d Array
class Solution:
    def runningSum(self, nums: list[int]) -> list[int]:
        for i in range(1, len(nums)):
            nums[i] += nums[i-1]
        return nums

List = [1,2,3,4]
sol = Solution()
print(sol.runningSum(List))
```

```python
#Shuffle the Array
class Solution:
    def shuffle(self, nums: list[int], n: int) -> list[int]:
        result = []
        for i in range(n):
            result.append(nums[i])      
            result.append(nums[i + n])  
        return result

List = [2,5,1,3,4,7]
n = 3
sol = Solution()
print(sol.shuffle(List, n))
```

```python
#Kids With the Greatest Number of Candies
class Solution:
    def kidsWithCandies(self, candies: list[int], extraCandies: int) -> list[bool]:
        max_candies = max(candies)
        result = []
        
        for c in candies:
            result.append(c + extraCandies >= max_candies)
        
        return result

candies = [2,3,5,1,3]
extraCandies = 3
sol = Solution()
print(sol.kidsWithCandies(candies, extraCandies))
```

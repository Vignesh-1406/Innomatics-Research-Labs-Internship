# Analysis on ML Test Scores

## Batch-wise Score Analysis

**Objective:**  
To analyze and derive insights from score data of three batches  
- AI_ELITE_7  
- AI_ELITE_6  
- AI_ELITE_4  

Each student is evaluated on a score out of 7.

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("scores_data.csv")
df.head()
```

## Dataset Description

- **Batch**: Represents the batch name  
- **User_ID**: Unique identifier for each student  
- **Score**: Score obtained out of 7  

The dataset is used to analyze performance trends and variations across batches.


```python
df.info()
df.describe()
```

```python
df.columns = df.columns.str.strip()
df.columns
```

```python
df["Score"] = df["Score"].astype(str).str.split("/").str[0].astype(int)
```

```python
df.head()
df.info()
```

```python
df.groupby("Batch")["Score"].describe()
```

## Batch-wise Insights

**AI_ELITE_7**
- Achieved the highest average scores across all batches
- Majority of students secured scores of 6 or 7
- Reflects strong consistency and solid subject mastery

**AI_ELITE_6**
- Demonstrates moderate overall performance
- Scores vary widely from very low to very high
- Highlights significant variability in student outcomes

**AI_ELITE_4**
- Records comparatively lower average scores than other batches
- Limited number of high achievers
- Indicates the need for focused foundational support


```python
df.groupby("Batch")["Score"].mean().plot(kind="bar")
plt.title("Average Score by Batch")
plt.xlabel("Batch")
plt.ylabel("Average Score")
plt.show()
```

```python
df["Score"].plot(kind="hist", bins=7)
plt.title("Overall Score Distribution")
plt.xlabel("Score")
plt.ylabel("Number of Students")
plt.show()
```

## Comparative Analysis

- **AI_ELITE_7** exhibits strong and consistently high performance.
- **AI_ELITE_6** reflects varied outcomes, indicating the need for selective guidance and mentoring.
- **AI_ELITE_4** displays comparatively weaker results, requiring concentrated academic assistance.


## Conclusion

The analysis highlights distinct performance variations across batches.
These trends reveal unique strengths and learning challenges.
Customized teaching and targeted interventions can enhance outcomes for each group.
